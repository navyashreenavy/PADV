```python
# ============================================
# PANDAS I/O FUNCTIONS - GOOGLE COLAB
# ============================================

import pandas as pd
import sqlite3

print("Pandas Version:", pd.__version__)
print("=" * 50)


# ============================================
# 1. pd.read_csv()
# ============================================

print("\n1. pd.read_csv()")
print("-" * 30)

# Create a sample CSV file
csv_data = """Name,Age,Marks
Rahul,20,85
Priya,21,90
Arun,19,78
"""

with open("students.csv", "w") as file:
    file.write(csv_data)

# Read CSV file
df_csv = pd.read_csv("students.csv")

print(df_csv)


# ============================================
# 2. pd.read_json()
# ============================================

print("\n2. pd.read_json()")
print("-" * 30)

# Create a sample JSON file
json_data = """
[
    {"Name": "Rahul", "Age": 20, "Marks": 85},
    {"Name": "Priya", "Age": 21, "Marks": 90},
    {"Name": "Arun", "Age": 19, "Marks": 78}
]
"""

with open("students.json", "w") as file:
    file.write(json_data)

# Read JSON file
df_json = pd.read_json("students.json")

print(df_json)


# ============================================
# 3. pd.read_html()
# ============================================

print("\n3. pd.read_html()")
print("-" * 30)

# Create a sample HTML file containing a table
html_data = """
<html>
<body>

<table>
    <tr>
        <th>Name</th>
        <th>Age</th>
        <th>Marks</th>
    </tr>

    <tr>
        <td>Rahul</td>
        <td>20</td>
        <td>85</td>
    </tr>

    <tr>
        <td>Priya</td>
        <td>21</td>
        <td>90</td>
    </tr>

    <tr>
        <td>Arun</td>
        <td>19</td>
        <td>78</td>
    </tr>
</table>

</body>
</html>
"""

with open("students.html", "w") as file:
    file.write(html_data)

# Read HTML table
tables = pd.read_html("students.html")

# read_html() returns a list of DataFrames
df_html = tables[0]

print(df_html)


# ============================================
# 4. pd.read_xml()
# ============================================

print("\n4. pd.read_xml()")
print("-" * 30)

# Create a sample XML file
xml_data = """
<students>
    <student>
        <Name>Rahul</Name>
        <Age>20</Age>
        <Marks>85</Marks>
    </student>

    <student>
        <Name>Priya</Name>
        <Age>21</Age>
        <Marks>90</Marks>
    </student>

    <student>
        <Name>Arun</Name>
        <Age>19</Age>
        <Marks>78</Marks>
    </student>
</students>
"""

with open("students.xml", "w") as file:
    file.write(xml_data)

# Read XML file
df_xml = pd.read_xml("students.xml")

print(df_xml)


# ============================================
# 5. pd.read_sql()
# ============================================

print("\n5. pd.read_sql()")
print("-" * 30)

# Create SQLite database
connection = sqlite3.connect("students.db")

# Create a DataFrame
data = {
    "Name": ["Rahul", "Priya", "Arun"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 78]
}

df_students = pd.DataFrame(data)

# Save DataFrame into SQL table
df_students.to_sql(
    "students",
    connection,
    if_exists="replace",
    index=False
)

# SQL query
query = "SELECT * FROM students"

# Read SQL table using Pandas
df_sql = pd.read_sql(query, connection)

print(df_sql)

# Close database connection
connection.close()


# ============================================
# FINAL SUMMARY
# ============================================

print("\n" + "=" * 50)
print("ALL FIVE PANDAS I/O FUNCTIONS EXECUTED!")
print("=" * 50)

print("""
1. pd.read_csv()  → Reads CSV files
2. pd.read_json() → Reads JSON files
3. pd.read_html() → Reads HTML tables
4. pd.read_xml()  → Reads XML files
5. pd.read_sql()  → Reads SQL database tables
""")
```
