```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

```python
import pandas as pd
import numpy as np

# Create a mock dataset representing AQI data
np.random.seed(42)
months = ['January', 'February', 'March', 'April', 'May', 'June']
cities = ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix']

data = {
    'Month': np.random.choice(months, size=200),
    'City': np.random.choice(cities, size=200),
    'AQI': np.random.randint(20, 150, size=200),
    'PM25': np.random.uniform(5, 55, size=200),
    'PM10': np.random.uniform(10, 100, size=200),
    'NO2': np.random.uniform(5, 40, size=200)
}

df = pd.DataFrame(data)
display(df.head())
```

```python
month_order = [
    'January', 'February', 'March',
    'April', 'May', 'June'
]
```

```python
monthly_aqi = (
    df.groupby('Month')['AQI']
    .mean()
    .reindex(month_order)
)

city_aqi = df.groupby('City')['AQI'].mean()

pollutant_data = df.groupby('City')[
    ['PM25', 'PM10', 'NO2']
].mean()

# Display calculated data to verify it works
print("Monthly AQI:")
print(monthly_aqi)
print("\nCity AQI:")
print(city_aqi)
```

```python
fig = make_subplots(
    rows=3,
    cols=2,
    specs=[
        [{'type': 'indicator'}, {'type': 'indicator'}],
        [{'type': 'xy'}, {'type': 'xy'}],
        [{'type': 'xy'}, {'type': 'xy'}]
    ],
    subplot_titles=[
        'Overall Average AQI',
        'Total Records',
        'Monthly AQI Trend',
        'Average AQI by City',
        'Pollutant Comparison',
        'PM2.5 vs AQI'
    ],
    vertical_spacing=0.12,
    horizontal_spacing=0.12
)
```

```python
# 1. Calculate metrics and prepare data structures
average_aqi = df['AQI'].mean()
total_records = len(df)

# Group data for plots
monthly_data = df.groupby('Month')['AQI'].mean().reindex(month_order).reset_index()
city_data = df.groupby('City')['AQI'].mean().reset_index()
pollutants = df.groupby('City')[['PM25', 'PM10', 'NO2']].mean().reset_index()

# 2. Configure subplot layout
fig = make_subplots(
    rows=3,
    cols=2,
    specs=[
        [{'type': 'indicator'}, {'type': 'indicator'}],
        [{'type': 'xy'}, {'type': 'xy'}],
        [{'type': 'xy'}, {'type': 'xy'}]
    ],
    subplot_titles=[
        'Overall Average AQI',
        'Total Records',
        'Monthly AQI Trend',
        'Average AQI by City',
        'Pollutant Comparison',
        'PM2.5 vs AQI'
    ],
    vertical_spacing=0.12,
    horizontal_spacing=0.12
)

# Subplot 1 (Row 1, Col 1): Overall Average AQI
fig.add_trace(
    go.Indicator(
        mode='number',
        value=average_aqi,
        number={'valueformat': '.2f'}
    ),
    row=1, col=1
)

# Subplot 2 (Row 1, Col 2): Total Records
fig.add_trace(
    go.Indicator(
        mode='number',
        value=total_records
    ),
    row=1, col=2
)

# Subplot 3 (Row 2, Col 1): Monthly AQI Trend
fig.add_trace(
    go.Scatter(
        x=monthly_data['Month'],
        y=monthly_data['AQI'],
        mode='lines+markers',
        name='AQI Trend'
    ),
    row=2, col=1
)

# Subplot 4 (Row 2, Col 2): Average AQI by City
fig.add_trace(
    go.Bar(
        x=city_data['City'],
        y=city_data['AQI'],
        name='City AQI'
    ),
    row=2, col=2
)

# Subplot 5 (Row 3, Col 1): Pollutant Comparison
fig.add_trace(go.Bar(x=pollutants['City'], y=pollutants['PM25'], name='PM2.5'), row=3, col=1)
fig.add_trace(go.Bar(x=pollutants['City'], y=pollutants['PM10'], name='PM10'), row=3, col=1)
fig.add_trace(go.Bar(x=pollutants['City'], y=pollutants['NO2'], name='NO2'), row=3, col=1)

# Subplot 6 (Row 3, Col 2): PM2.5 vs AQI
fig.add_trace(
    go.Scatter(
        x=df['PM25'],
        y=df['AQI'],
        mode='markers',
        name='PM2.5 vs AQI',
        marker=dict(opacity=0.6)
    ),
    row=3, col=2
)

# 3. Adjust layout settings
fig.update_layout(
    height=900,
    width=950,
    showlegend=True,
    title_text="Air Quality Dashboard Overview",
    title_x=0.5
)
fig.show()
```
